```python
import pandas as pd, os, shutil

cols = ["year_month", "year", "season", "category", "channel", "store_city", "has_promo",
        "total_value", "estimated_profit", "real_profit", "quantity"]
d = pd.read_parquet(r"D:\retailpulse\retailpulse_sales_clean.parquet", columns=cols)

cube = (d.groupby(["year_month", "year", "season", "category", "channel", "store_city", "has_promo"], observed=True)
          .agg(sales=("total_value", "sum"), profit_30=("estimated_profit", "sum"),
               real_profit=("real_profit", "sum"), units=("quantity", "sum"), lines=("quantity", "size"))
          .reset_index().round(2))
path = r"D:\retailpulse\app_data\sales_cube.csv"
cube.to_csv(path, index=False)
shutil.copy(r"D:\retailpulse\customer_table.csv", r"D:\retailpulse\app_data\customer_table.csv")

print(cube.shape, round(os.path.getsize(path) / 1e6, 1), "MB")
```
