```python
path = r"D:\retailpulse\datarawsynthetic\retail_contaminated_dataset\sales_transactions.csv"
rows = 0; nulls = None; dups = 0
bad_qty = neg_qty = bad_price = 0
channels = pd.Series(dtype="float64"); date_len = pd.Series(dtype="float64")

for chunk in pd.read_csv(path, chunksize=1_000_000, dtype=str):
    rows += len(chunk)
    n = chunk.isna().sum()
    nulls = n if nulls is None else nulls + n
    dups += chunk.duplicated().sum()
    q = pd.to_numeric(chunk["quantity"], errors="coerce")
    p = pd.to_numeric(chunk["unit_price"], errors="coerce")
    bad_qty += q.isna().sum() - chunk["quantity"].isna().sum()   # text in a number column
    neg_qty += (q <= 0).sum()
    bad_price += (p.isna() | (p <= 0)).sum()
    channels = channels.add(chunk["channel"].value_counts(dropna=False), fill_value=0)
    date_len = date_len.add(chunk["date"].str.len().value_counts(dropna=False), fill_value=0)

print("rows:", rows)
print("missing values:\n", nulls)
print("duplicate rows (within chunks):", dups)
print("non-numeric quantity:", bad_qty, "| quantity <= 0:", neg_qty, "| bad price:", bad_price)
print("channel values:\n", channels)
print("date string lengths:\n", date_len)
```

```python
base  = r"D:\retailpulse\datarawsynthetic"
dirty = base + r"\retail_contaminated_dataset"
clean = base + r"\retail_clean_dataset"
path  = dirty + r"\sales_transactions.csv"

# 1. how many rows does the clean answer key have?
clean_rows = 0
for ch in pd.read_csv(clean + r"\sales_transactions.csv", chunksize=1_000_000, usecols=["receipt_id"]):
    clean_rows += len(ch)
print("clean rows:", clean_rows, "| contaminated:", 9945511, "| extra rows:", 9945511 - clean_rows)

# 2. deeper checks on the contaminated file
skus   = set(pd.read_csv(dirty + r"\sku_master.csv")["sku_id"])
custs  = set(pd.read_csv(dirty + r"\customer_master.csv")["cust_id"])
stores = set(pd.read_csv(dirty + r"\store_master.csv")["store_id"])

s = dict(total_mismatch=0, disc_rows=0, disc_applied=0, disc_out_of_range=0,
         orphan_sku=0, orphan_cust=0, orphan_store=0)
dmin, dmax = "9999", "0000"
qmax = pmax = 0; pmin = 1e12

for ch in pd.read_csv(path, chunksize=1_000_000):
    dmin = min(dmin, ch["date"].min()); dmax = max(dmax, ch["date"].max())
    calc = ch["quantity"] * ch["unit_price"]
    s["total_mismatch"] += (abs(calc - ch["total_value"]) > 0.05).sum()
    d = ch["discount_pct"] > 0
    s["disc_rows"] += d.sum()
    s["disc_applied"] += (d & (abs(calc * (1 - ch["discount_pct"]/100) - ch["total_value"]) <= 0.05)).sum()
    s["disc_out_of_range"] += ((ch["discount_pct"] < 0) | (ch["discount_pct"] > 100)).sum()
    s["orphan_sku"]   += (~ch["sku_id"].isin(skus)).sum()
    s["orphan_cust"]  += (~ch["customer_id"].isin(custs)).sum()
    s["orphan_store"] += (~ch["store_id"].isin(stores)).sum()
    qmax = max(qmax, ch["quantity"].max())
    pmax = max(pmax, ch["unit_price"].max()); pmin = min(pmin, ch["unit_price"].min())

print(s)
print("date range:", dmin, "to", dmax)
print("max quantity:", qmax, "| unit_price range:", pmin, "to", pmax)
```

```python
dirty = r"D:\retailpulse\datarawsynthetic\retail_contaminated_dataset"
clean = r"D:\retailpulse\datarawsynthetic\retail_clean_dataset"

for f in ["customer_master", "sku_master", "store_master",
          "inventory_snapshot", "sku_inventory_flags", "promotions"]:
    d = pd.read_csv(f"{dirty}\\{f}.csv")
    c = pd.read_csv(f"{clean}\\{f}.csv")
    print("=====", f, "| dirty", d.shape, "| clean", c.shape)
    print("missing (dirty):", d.isna().sum()[d.isna().sum() > 0].to_dict())
    print("missing (clean):", c.isna().sum()[c.isna().sum() > 0].to_dict())
    print("duplicate rows -> dirty:", d.duplicated().sum(), "| clean:", c.duplicated().sum())
    for name, t in [("dirty", d), ("clean", c)]:
        num = t.select_dtypes("number")
        if not num.empty:
            print(name, "min/max:\n", num.agg(["min", "max"]).T.to_string())
    for col in d.select_dtypes("object"):
        if d[col].nunique() <= 12:
            print("dirty", col, d[col].value_counts(dropna=False).to_dict())
    print()
```

```python
sales = pd.read_csv(dirty + r"\sales_transactions.csv", nrows=1_000_000)
sku = pd.read_csv(dirty + r"\sku_master.csv")[["sku_id", "unit_price", "cost_price"]]
m = sales.merge(sku, on="sku_id", suffixes=("", "_list"))
print("price differs from list price:", (abs(m["unit_price"] - m["unit_price_list"]) > 0.05).sum(), "of", len(m))
print("sold below cost:", (m["unit_price"] < m["cost_price"]).sum())
```

```python
inv_d = pd.read_csv(dirty + r"\inventory_snapshot.csv")
inv_c = pd.read_csv(clean + r"\inventory_snapshot.csv")
fl_d  = pd.read_csv(dirty + r"\sku_inventory_flags.csv")

print(list(inv_d.columns))
print(inv_d.head(3).to_string())
print(list(fl_d.columns))
print(fl_d[fl_d["flag"] == "STOCKOUT_RISK"].head(3).to_string())

key = [c for c in ["sku_id", "store_id"] if c in inv_d.columns]
print("key:", key)
print("repeated key rows -> dirty:", inv_d.duplicated(key).sum(), "| clean:", inv_c.duplicated(key).sum())
```

```python
win = fl_d[fl_d["flag"] == "STOCKOUT_RISK"][["sku_id", "window_start", "window_end"]]
win_skus = set(win["sku_id"])

def load(folder):
    parts = []
    for ch in pd.read_csv(folder + r"\sales_transactions.csv", chunksize=1_000_000,
                          usecols=["date", "sku_id", "total_value"]):
        parts.append(ch[ch["sku_id"].isin(win_skus)])
    return pd.concat(parts)

def in_window(s):
    m = s.merge(win, on="sku_id")
    return m[(m["date"] >= m["window_start"]) & (m["date"] <= m["window_end"])]

sd, sc = load(dirty), load(clean)
print("rows for these 200 SKUs -> dirty:", len(sd), "| clean:", len(sc))
print("rows inside their stockout windows -> dirty:", len(in_window(sd)), "| clean:", len(in_window(sc)))
```
